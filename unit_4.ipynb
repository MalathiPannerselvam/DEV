```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from ucimlrepo import fetch_ucirepo

student_performance = fetch_ucirepo(id=320)

X = student_performance.data.features
y = student_performance.data.targets

df = pd.concat([X, y], axis=1)

print("DATASET LOADED SUCCESSFULLY")
print("Shape:", df.shape)


# Numerical - Numerical Analysis

pearson_corr = df["G2"].corr(
    df["G3"],
    method="pearson"
)

spearman_corr = df["G2"].corr(
    df["G3"],
    method="spearman"
)

print("\nPEARSON CORRELATION")
print(pearson_corr)

print("\nSPEARMAN CORRELATION")
print(spearman_corr)


# Scatter Plot

plt.figure(figsize=(8, 5))

plt.scatter(
    df["G2"],
    df["G3"],
    alpha=0.6
)

plt.xlabel("G2")
plt.ylabel("G3")
plt.title("G2 vs G3")

plt.grid(True)
plt.show()


# Scatter Plot with Regression Line

x = df["G2"].to_numpy()
y_values = df["G3"].to_numpy()

slope, intercept = np.polyfit(
    x,
    y_values,
    1
)

predicted = slope * x + intercept

plt.figure(figsize=(8, 5))

plt.scatter(
    x,
    y_values,
    alpha=0.5
)

plt.plot(
    x,
    predicted,
    linewidth=2
)

plt.xlabel("G2")
plt.ylabel("G3")
plt.title("G2 vs G3 with Regression Line")

plt.show()


# Categorical - Categorical Analysis

gender_internet = pd.crosstab(
    df["sex"],
    df["internet"]
)

print("\nGENDER VS INTERNET")
print(gender_internet)


# Grouped Bar Chart

gender_internet.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.title("Internet Access by Gender")

plt.xticks(rotation=0)
plt.legend(title="Internet")

plt.show()


# Stacked Bar Chart

gender_internet.plot(
    kind="bar",
    stacked=True,
    figsize=(8, 5)
)

plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.title("Stacked Internet Access by Gender")

plt.xticks(rotation=0)
plt.legend(title="Internet")

plt.show()


# Numerical - Categorical Analysis

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="sex",
    y="G3"
)

plt.xlabel("Gender")
plt.ylabel("Final Grade G3")
plt.title("G3 Distribution by Gender")

plt.show()


# Violin Plot

plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df,
    x="sex",
    y="G3"
)

plt.xlabel("Gender")
plt.ylabel("Final Grade G3")
plt.title("G3 Distribution by Gender - Violin Plot")

plt.show()


# Box Plot by Study Time

plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="studytime",
    y="G3"
)

plt.xlabel("Study Time")
plt.ylabel("Final Grade G3")
plt.title("Final Grade by Study Time")

plt.show()


# Violin Plot by Study Time

plt.figure(figsize=(9, 5))

sns.violinplot(
    data=df,
    x="studytime",
    y="G3"
)

plt.xlabel("Study Time")
plt.ylabel("Final Grade G3")
plt.title("Final Grade by Study Time")

plt.show()


# Correlation Matrix

correlation = df[
    ["age", "studytime", "failures", "absences", "G1", "G2", "G3"]
].corr()

print("\nCORRELATION MATRIX")
print(correlation)


plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()


print("\nBIVARIATE ANALYSIS COMPLETED SUCCESSFULLY")
```
