```python
!pip -q install ucimlrepo plotly bokeh seaborn scikit-learn

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

from ucimlrepo import fetch_ucirepo
from bokeh.plotting import figure, show
from bokeh.io import output_notebook
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

output_notebook()
sns.set_theme(style="whitegrid")

# =========================================================
# LOAD DATASET
# =========================================================

student_performance = fetch_ucirepo(id=320)
X = student_performance.data.features
target = student_performance.data.targets
df = pd.concat([X, target], axis=1)

print("=" * 60)
print("EXPLORATORY DATA ANALYSIS AND VISUALIZATION")
print("OF STUDENT PERFORMANCE DATASET")
print("=" * 60)

print("Dataset Shape:", df.shape)
print("\nFirst Five Rows:")
display(df.head())

print("\nLast Five Rows:")
display(df.tail())

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

numerical_columns = df.select_dtypes(include=np.number).columns.tolist()
categorical_columns = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("\nNumerical Columns:", numerical_columns)
print("\nCategorical Columns:", categorical_columns)

print("\nDescriptive Statistics:")
display(df.describe(include="all").T)

# =========================================================
# UNIT 1: EXPLORATORY DATA ANALYSIS
# =========================================================

print("\nUNIT 1: EXPLORATORY DATA ANALYSIS")

print("\nSelected Columns:")
display(df[["age", "studytime", "G3"]].head())

high_performers = df[df["G3"] >= 15]
print("High Performers (G3 >= 15):", len(high_performers))

for col in numerical_columns:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].median())

for col in categorical_columns:
    if df[col].isnull().any():
        mode_value = df[col].mode()
        if not mode_value.empty:
            df[col] = df[col].fillna(mode_value.iloc[0])

df["sex_binary"] = df["sex"].map({"F": 0, "M": 1})

print("\nBinary Encoding:")
display(df[["sex", "sex_binary"]].head())

encoded_df = pd.get_dummies(
    df,
    columns=categorical_columns,
    drop_first=False,
    dtype=int
)

print("Shape After One-Hot Encoding:", encoded_df.shape)

df["log_absences"] = np.log1p(df["absences"])
df["G3_standardized"] = (
    (df["G3"] - df["G3"].mean()) / df["G3"].std()
)

print("\nTransformed Data:")
display(df[["absences", "log_absences", "G3", "G3_standardized"]].head())

print("\nAverage Grade by Study Time:")
display(df.groupby("studytime")["G3"].agg(
    ["count", "mean", "median", "std"]
).round(3))

print("\nAverage Grade by Gender:")
display(df.groupby("sex")["G3"].agg(
    ["count", "mean", "median", "std"]
).round(3))

# =========================================================
# UNIT 2: DATA VISUALIZATION
# =========================================================

print("\nUNIT 2: DATA VISUALIZATION")

mean_g3_age = df.groupby("age")["G3"].mean()

# Line Plot
plt.figure(figsize=(8, 5))
plt.plot(mean_g3_age.index, mean_g3_age.values, marker="o",
         label="Average G3")
plt.xlabel("Age")
plt.ylabel("Average Final Grade")
plt.title("Average Final Grade by Age")
plt.legend()
plt.grid(True)
plt.show()

# Scatter Plot
plt.figure(figsize=(8, 5))
plt.scatter(df["G2"], df["G3"], alpha=0.6)
plt.xlabel("G2")
plt.ylabel("G3")
plt.title("G2 vs G3")
plt.grid(True)
plt.show()

# Error Bar Plot
plt.figure(figsize=(6, 5))
plt.errorbar(["G3"], [df["G3"].mean()],
             yerr=[df["G3"].std()], fmt="o", capsize=8)
plt.ylabel("Grade")
plt.title("Mean Final Grade with Standard Deviation")
plt.show()

# Histogram
plt.figure(figsize=(8, 5))
plt.hist(df["G3"], bins=10, edgecolor="black")
plt.xlabel("Final Grade")
plt.ylabel("Frequency")
plt.title("Distribution of Final Grades")
plt.show()

# Density Plot
values = df["G3"].dropna().to_numpy()
grid = np.linspace(values.min(), values.max(), 250)
bandwidth = max(1.06 * values.std() * len(values) ** (-1 / 5), 0.1)

density = np.zeros_like(grid, dtype=float)

for value in values:
    density += np.exp(-0.5 * ((grid - value) / bandwidth) ** 2)

density /= len(values) * bandwidth * np.sqrt(2 * np.pi)

plt.figure(figsize=(8, 5))
plt.plot(grid, density)
plt.fill_between(grid, density, alpha=0.25)
plt.xlabel("Final Grade")
plt.ylabel("Density")
plt.title("Density Plot of Final Grades")
plt.show()

# Contour-Style Plot
plt.figure(figsize=(8, 5))
plt.hist2d(df["G2"], df["G3"], bins=15)
plt.colorbar(label="Frequency")
plt.xlabel("G2")
plt.ylabel("G3")
plt.title("Density Visualization of G2 and G3")
plt.show()

# Subplots
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].hist(df["G3"], bins=10, edgecolor="black")
axes[0].set_title("Final Grade Distribution")
axes[0].set_xlabel("G3")
axes[0].set_ylabel("Frequency")

axes[1].scatter(df["G2"], df["G3"], alpha=0.6)
axes[1].set_title("G2 vs G3")
axes[1].set_xlabel("G2")
axes[1].set_ylabel("G3")

plt.tight_layout()
plt.show()

# Annotation
plt.figure(figsize=(8, 5))
plt.scatter(df["G2"], df["G3"], alpha=0.6)
plt.annotate(
    "High Final Grade",
    xy=(17, 18),
    xytext=(12, 18),
    arrowprops={"arrowstyle": "->"}
)
plt.xlabel("G2")
plt.ylabel("G3")
plt.title("G2 vs G3 with Annotation")
plt.show()

# Seaborn Histogram
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="G3", bins=10, kde=True)
plt.title("Seaborn Histogram of Final Grades")
plt.show()

# Seaborn Boxplot
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="sex", y="G3")
plt.title("Final Grade by Gender")
plt.show()

# Plotly Interactive Scatter Plot
fig = px.scatter(
    df,
    x="G2",
    y="G3",
    color="sex",
    hover_data=["age", "studytime", "absences"],
    title="Interactive G2 vs G3"
)
fig.show()

# Plotly 3D Plot
fig = px.scatter_3d(
    df,
    x="studytime",
    y="G2",
    z="G3",
    color="sex",
    title="3D Student Performance Visualization"
)
fig.show()

# Bokeh Plot
bokeh_fig = figure(
    title="Bokeh: G2 vs G3",
    x_axis_label="G2",
    y_axis_label="G3",
    width=700,
    height=450
)

bokeh_fig.scatter(df["G2"], df["G3"], size=7, alpha=0.5)
show(bokeh_fig)

# =========================================================
# UNIT 3: UNIVARIATE ANALYSIS
# =========================================================

print("\nUNIT 3: UNIVARIATE ANALYSIS")

uni_cols = [
    "age", "studytime", "failures",
    "absences", "G1", "G2", "G3"
]

print("\nSummary Statistics:")
display(df[uni_cols].describe().T)

print("\nMean:")
print(df[uni_cols].mean())

print("\nMedian:")
print(df[uni_cols].median())

print("\nStandard Deviation:")
print(df[uni_cols].std())

print("\nQuartiles:")
print(df[uni_cols].quantile([0.25, 0.50, 0.75]))

print("\nSkewness:")
print(df[uni_cols].skew())

# Outlier Detection Using IQR
q1 = df["G3"].quantile(0.25)
q3 = df["G3"].quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = df[
    (df["G3"] < lower_bound) |
    (df["G3"] > upper_bound)
]

print("\nG3 Outlier Analysis")
print("Q1:", q1)
print("Median:", df["G3"].median())
print("Q3:", q3)
print("IQR:", iqr)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))

# Box Plot
plt.figure(figsize=(7, 5))
plt.boxplot(df["G3"])
plt.ylabel("Final Grade")
plt.title("Box Plot of Final Grades")
plt.show()

# Gender Frequency
print("\nGender Frequency:")
print(df["sex"].value_counts())

print("\nSchool Frequency:")
print(df["school"].value_counts())

# Bar Chart
df["sex"].value_counts().plot(kind="bar", figsize=(6, 4))
plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.title("Students by Gender")
plt.xticks(rotation=0)
plt.show()

# Pie Chart
df["sex"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(6, 6)
)
plt.title("Gender Distribution")
plt.ylabel("")
plt.show()

# =========================================================
# UNIT 4: BIVARIATE ANALYSIS
# =========================================================

print("\nUNIT 4: BIVARIATE ANALYSIS")

pearson_corr = df["G2"].corr(df["G3"], method="pearson")
spearman_corr = df["G2"].corr(df["G3"], method="spearman")

print("Pearson Correlation (G2 and G3):", pearson_corr)
print("Spearman Correlation (G2 and G3):", spearman_corr)

# Regression Line
x_values = df["G2"].to_numpy()
y_values = df["G3"].to_numpy()

slope, intercept = np.polyfit(x_values, y_values, 1)

plt.figure(figsize=(8, 5))
plt.scatter(x_values, y_values, alpha=0.5, label="Students")
plt.plot(
    x_values,
    slope * x_values + intercept,
    label="Regression Line"
)
plt.xlabel("G2")
plt.ylabel("G3")
plt.title("G2 vs G3 with Regression Line")
plt.legend()
plt.show()

print("Regression Slope:", slope)
print("Regression Intercept:", intercept)

# Categorical vs Categorical
gender_internet = pd.crosstab(df["sex"], df["internet"])

print("\nGender vs Internet Access:")
display(gender_internet)

gender_internet.plot(kind="bar", figsize=(7, 4))
plt.title("Internet Access by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.show()

gender_internet.plot(kind="bar", stacked=True, figsize=(7, 4))
plt.title("Stacked Internet Access by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.show()

# Numerical vs Categorical: Box Plot
plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x="sex", y="G3")
plt.title("Final Grade by Gender")
plt.show()

# Violin Plot
plt.figure(figsize=(7, 4))
sns.violinplot(data=df, x="sex", y="G3")
plt.title("Final Grade Distribution by Gender")
plt.show()

# Grade by Study Time
plt.figure(figsize=(8, 4))
sns.boxplot(data=df, x="studytime", y="G3")
plt.title("Final Grade by Study Time")
plt.show()

# Correlation Heatmap
correlation = df[uni_cols].corr()

plt.figure(figsize=(9, 7))
sns.heatmap(correlation, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()

# =========================================================
# UNIT 5: MULTIVARIATE ANALYSIS AND MACHINE LEARNING
# =========================================================

print("\nUNIT 5: MULTIVARIATE ANALYSIS AND MACHINE LEARNING")

# Pair Plot
sns.pairplot(df[uni_cols])
plt.suptitle("Pair Plot of Student Performance Variables", y=1.02)
plt.show()

# Grouped Analysis
grouped = (
    df.groupby(["sex", "studytime"])["G3"]
    .mean()
    .reset_index()
)

print("\nAverage G3 by Gender and Study Time:")
display(grouped)

plt.figure(figsize=(8, 5))
sns.barplot(data=grouped, x="studytime", y="G3", hue="sex")
plt.xlabel("Study Time")
plt.ylabel("Average Final Grade")
plt.title("Average Grade by Gender and Study Time")
plt.show()

# PCA
pca_features = [
    "age", "studytime", "failures",
    "absences", "G1", "G2", "G3"
]

scaler_pca = StandardScaler()
scaled_data = scaler_pca.fit_transform(df[pca_features])

pca = PCA(n_components=2)
pca_data = pca.fit_transform(scaled_data)

pca_df = pd.DataFrame(pca_data, columns=["PC1", "PC2"])
pca_df["sex"] = df["sex"].to_numpy()

print("\nPCA Explained Variance Ratio:")
print(pca.explained_variance_ratio_)

print(
    "Total Variance Explained by Two Components:",
    pca.explained_variance_ratio_.sum()
)

plt.figure(figsize=(8, 5))
sns.scatterplot(data=pca_df, x="PC1", y="PC2", hue="sex")
plt.title("PCA of Student Performance")
plt.show()

# Machine Learning
features = [
    "age", "studytime", "failures",
    "absences", "G1", "G2"
]

X_ml = df[features]
y_ml = df["G3"]

X_train, X_test, y_train, y_test = train_test_split(
    X_ml,
    y_ml,
    test_size=0.2,
    random_state=42
)

print("\nTraining Records:", len(X_train))
print("Testing Records:", len(X_test))

linear_model = LinearRegression()
linear_model.fit(X_train, y_train)
linear_predictions = linear_model.predict(X_test)

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)
rf_model.fit(X_train, y_train)
rf_predictions = rf_model.predict(X_test)

# Model Evaluation
results = []

for name, predictions in [
    ("Linear Regression", linear_predictions),
    ("Random Forest", rf_predictions)
]:
    mse = mean_squared_error(y_test, predictions)

    results.append({
        "Model": name,
        "MAE": mean_absolute_error(y_test, predictions),
        "MSE": mse,
        "RMSE": np.sqrt(mse),
        "R2 Score": r2_score(y_test, predictions)
    })

results_df = pd.DataFrame(results)

print("\nModel Comparison:")
display(results_df.round(4))

# Feature Importance
importance_df = pd.DataFrame({
    "Feature": features,
    "Importance": rf_model.feature_importances_
}).sort_values("Importance", ascending=False)

print("\nRandom Forest Feature Importance:")
display(importance_df)

plt.figure(figsize=(8, 5))
sns.barplot(data=importance_df, x="Importance", y="Feature")
plt.title("Feature Importance for Final Grade Prediction")
plt.show()

# Actual vs Predicted
plt.figure(figsize=(7, 5))
plt.scatter(y_test, rf_predictions, alpha=0.7)

plt.xlabel("Actual G3")
plt.ylabel("Predicted G3")
plt.title("Actual vs Predicted Final Grades")

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    linestyle="--"
)

plt.show()

# Example Prediction
new_student = pd.DataFrame([{
    "age": 16,
    "studytime": 2,
    "failures": 0,
    "absences": 2,
    "G1": 14,
    "G2": 15
}])

prediction = rf_model.predict(new_student)

print("\nExample Student Details:")
display(new_student)

print("Predicted Final Grade (G3):", round(float(prediction[0]), 2))

# =========================================================
# KEY FINDINGS AND ETHICAL CONSIDERATIONS
# =========================================================

print("\nKEY FINDINGS")
print("1. Dataset records:", len(df))
print("2. Original dataset columns:", len(X.columns) + target.shape[1])
print("3. Pearson correlation between G2 and G3:", round(pearson_corr, 3))
print("4. Students with G3 >= 15:", int((df["G3"] >= 15).sum()))
print(
    "5. PCA variance explained:",
    round(pca.explained_variance_ratio_.sum() * 100, 2),
    "%"
)

print("\nETHICAL CONSIDERATIONS")
print("1. Protect student privacy.")
print("2. Check for bias across student groups.")
print("3. Use predictions to support teachers, not replace them.")
print("4. Correlation does not prove causation.")
print("5. G1 and G2 are strong predictors but may not always be available.")

```
