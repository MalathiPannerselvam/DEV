```python
!pip -q install ucimlrepo

import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from ucimlrepo import fetch_ucirepo

student_performance = fetch_ucirepo(id=320)

X_original = student_performance.data.features
y_original = student_performance.data.targets

df = pd.concat(
    [X_original, y_original],
    axis=1
)

print("DATASET LOADED SUCCESSFULLY")
print("=" * 60)

print("\nDATASET SHAPE")
print(df.shape)

print("\nCOLUMN NAMES")
print(df.columns.tolist())

print("\nFIRST 5 ROWS")
display(df.head())

print("\nLAST 5 ROWS")
display(df.tail())

print("\nDATA TYPES")
print(df.dtypes)

print("\nDATASET INFORMATION")
df.info()

print("\nMISSING VALUES")
missing_values = df.isnull().sum()
display(missing_values)

print("Total Missing Values:", df.isnull().sum().sum())

print("\nDUPLICATE VALUES")
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)

if duplicate_count > 0:
    df = df.drop_duplicates()
    print("Duplicate rows removed.")
else:
    print("No duplicate rows found.")

print("\nNUMERICAL VARIABLES")
numerical_columns = df.select_dtypes(
    include=np.number
).columns.tolist()
print(numerical_columns)

print("\nCATEGORICAL VARIABLES")
categorical_columns = df.select_dtypes(
    exclude=np.number
).columns.tolist()
print(categorical_columns)

print("\nDATASET SLICING")

print("\nFirst 10 rows:")
display(df.iloc[:10])

print("\nSelected columns:")
display(
    df[
        [
            "age",
            "studytime",
            "failures",
            "absences",
            "G1",
            "G2",
            "G3"
        ]
    ].head()
)

print("\nStudents with G3 >= 15")

high_performers = df[df["G3"] >= 15]

print("Number of high performers:", len(high_performers))

display(high_performers.head())

print("\nBINARY ENCODING")

df_encoded = df.copy()

binary_mapping = {
    "yes": 1,
    "no": 0,
    "F": 0,
    "M": 1
}

for col in df_encoded.columns:

    if df_encoded[col].dtype == "object":

        unique_values = set(
            df_encoded[col].dropna().unique()
        )

        if unique_values.issubset(
            set(binary_mapping.keys())
        ):

            df_encoded[col] = (
                df_encoded[col].map(binary_mapping)
            )

print("Binary encoded dataset:")
display(df_encoded.head())

print("\nONE-HOT ENCODING")

categorical_for_encoding = (
    df.select_dtypes(
        exclude=np.number
    ).columns.tolist()
)

df_onehot = pd.get_dummies(
    df,
    columns=categorical_for_encoding,
    drop_first=True
)

print("Original shape:", df.shape)
print("Encoded shape:", df_onehot.shape)

display(df_onehot.head())

print("\nDATA TRANSFORMATION")

df_transformed = df.copy()

df_transformed["log_absences"] = (
    np.log1p(
        df_transformed["absences"]
    )
)

scaler = StandardScaler()

df_transformed["G3_standardized"] = (
    scaler.fit_transform(
        df_transformed[["G3"]]
    )
)

display(
    df_transformed[
        [
            "absences",
            "log_absences",
            "G3",
            "G3_standardized"
        ]
    ].head()
)

print("\nGROUPING DATASET")

study_group = (
    df.groupby("studytime")["G3"]
    .agg(
        [
            "count",
            "mean",
            "median",
            "std"
        ]
    )
)

print("G3 Analysis by Study Time:")
display(study_group)

gender_group = (
    df.groupby("sex")["G3"]
    .agg(
        [
            "count",
            "mean",
            "median",
            "std"
        ]
    )
)

print("\nPerformance by Gender:")
display(gender_group)

print("\nDESCRIPTIVE STATISTICS")

display(
    df[
        [
            "age",
            "Medu",
            "Fedu",
            "traveltime",
            "studytime",
            "failures",
            "famrel",
            "freetime",
            "goout",
            "Dalc",
            "Walc",
            "health",
            "absences",
            "G1",
            "G2",
            "G3"
        ]
    ].describe()
)

print("\nEDA COMPLETED SUCCESSFULLY")
```
