```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from ucimlrepo import fetch_ucirepo

student_performance = fetch_ucirepo(id=320)

X = student_performance.data.features
y = student_performance.data.targets

df = pd.concat([X, y], axis=1)

print("DATASET LOADED SUCCESSFULLY")
print("Shape:", df.shape)


# Numerical Variables
numerical_columns = [
    "age",
    "studytime",
    "failures",
    "absences",
    "G1",
    "G2",
    "G3"
]

print("\nNUMERICAL VARIABLES")
print(df[numerical_columns].describe())


# Summary Statistics
print("\nMEAN")
print(df[numerical_columns].mean())

print("\nMEDIAN")
print(df[numerical_columns].median())

print("\nSTANDARD DEVIATION")
print(df[numerical_columns].std())

print("\nQUARTILES")
print(df[numerical_columns].quantile([0.25, 0.50, 0.75]))


# Skewness
print("\nSKEWNESS")
print(df[numerical_columns].skew())


# G3 Quartiles
Q1 = df["G3"].quantile(0.25)
Q2 = df["G3"].quantile(0.50)
Q3 = df["G3"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["G3"] < lower_bound) |
    (df["G3"] > upper_bound)
]

print("\nG3 QUARTILES")
print("Q1:", Q1)
print("Median:", Q2)
print("Q3:", Q3)
print("IQR:", IQR)

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))


# Histogram
plt.figure(figsize=(8, 5))

plt.hist(
    df["G3"],
    bins=10,
    edgecolor="black"
)

plt.xlabel("Final Grade G3")
plt.ylabel("Frequency")
plt.title("Distribution of Final Grades")

plt.show()


# Box Plot
plt.figure(figsize=(8, 5))

plt.boxplot(
    df["G3"],
    vert=True
)

plt.ylabel("Final Grade G3")
plt.title("Box Plot of Final Grades")

plt.show()


# Density Plot without SciPy
values = df["G3"].dropna().to_numpy()

grid = np.linspace(
    values.min(),
    values.max(),
    200
)

bandwidth = 0.5

density = np.zeros_like(
    grid,
    dtype=float
)

for value in values:
    density += (
        np.exp(
            -0.5 * ((grid - value) / bandwidth) ** 2
        )
        / (bandwidth * np.sqrt(2 * np.pi))
    )

density /= len(values)

plt.figure(figsize=(8, 5))

plt.plot(
    grid,
    density,
    linewidth=2
)

plt.fill_between(
    grid,
    density,
    alpha=0.3
)

plt.xlabel("Final Grade G3")
plt.ylabel("Density")
plt.title("Density Plot of Final Grades")

plt.show()


# Categorical Variables
print("\nGENDER FREQUENCY")
print(df["sex"].value_counts())


print("\nSCHOOL FREQUENCY")
print(df["school"].value_counts())


# Frequency Table
gender_frequency = df["sex"].value_counts()

print("\nFREQUENCY TABLE")
print(gender_frequency)


# Bar Chart
plt.figure(figsize=(7, 5))

gender_frequency.plot(
    kind="bar"
)

plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.title("Students by Gender")

plt.xticks(rotation=0)

plt.show()


# Pie Chart
plt.figure(figsize=(7, 7))

gender_frequency.plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.ylabel("")
plt.title("Gender Distribution")

plt.show()


# Seaborn Bar Plot
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="sex"
)

plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.title("Gender Frequency")

plt.show()


# Seaborn Box Plot
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="sex",
    y="G3"
)

plt.xlabel("Gender")
plt.ylabel("Final Grade G3")
plt.title("Final Grade Distribution by Gender")

plt.show()


print("\nUNIVARIATE ANALYSIS COMPLETED SUCCESSFULLY")
```
