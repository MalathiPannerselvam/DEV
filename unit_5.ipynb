```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from ucimlrepo import fetch_ucirepo

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

student_performance = fetch_ucirepo(id=320)

X = student_performance.data.features
y = student_performance.data.targets

df = pd.concat([X, y], axis=1)

print("DATASET LOADED SUCCESSFULLY")
print("Shape:", df.shape)


# Selected numerical variables

selected_columns = [
    "age",
    "studytime",
    "failures",
    "absences",
    "G1",
    "G2",
    "G3"
]

selected_df = df[selected_columns]

print("\nSELECTED VARIABLES")
print(selected_df.head())


# Multivariate Correlation

correlation = selected_df.corr()

print("\nCORRELATION MATRIX")
print(correlation)


# Heatmap

plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Multivariate Correlation Heatmap")

plt.show()


# Pair Plot

sns.pairplot(
    selected_df[
        ["studytime", "failures", "absences", "G1", "G2", "G3"]
    ]
)

plt.show()


# Grouped Visualization

grouped_data = df.groupby(
    ["sex", "studytime"]
)["G3"].mean().reset_index()

print("\nAVERAGE G3 BY GENDER AND STUDY TIME")
print(grouped_data)


plt.figure(figsize=(9, 6))

sns.barplot(
    data=grouped_data,
    x="studytime",
    y="G3",
    hue="sex"
)

plt.xlabel("Study Time")
plt.ylabel("Average G3")
plt.title("Average Final Grade by Gender and Study Time")

plt.show()


# PCA

pca_columns = [
    "age",
    "studytime",
    "failures",
    "absences",
    "G1",
    "G2",
    "G3"
]

pca_data = df[pca_columns]

scaler = StandardScaler()

scaled_data = scaler.fit_transform(
    pca_data
)

pca = PCA(
    n_components=2
)

principal_components = pca.fit_transform(
    scaled_data
)

pca_df = pd.DataFrame(
    principal_components,
    columns=["PC1", "PC2"]
)

print("\nPCA EXPLAINED VARIANCE")
print(pca.explained_variance_ratio_)

print(
    "Total Variance Explained:",
    pca.explained_variance_ratio_.sum()
)


# PCA Visualization

plt.figure(figsize=(8, 6))

plt.scatter(
    pca_df["PC1"],
    pca_df["PC2"],
    alpha=0.6
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Visualization")

plt.grid(True)

plt.show()


# PCA Loadings

loadings = pd.DataFrame(
    pca.components_.T,
    columns=["PC1", "PC2"],
    index=pca_columns
)

print("\nPCA LOADINGS")
print(loadings)


# Machine Learning

features = [
    "age",
    "studytime",
    "failures",
    "absences",
    "G1",
    "G2"
]

target = "G3"

X_ml = df[features]
y_ml = df[target]

X_train, X_test, y_train, y_test = train_test_split(
    X_ml,
    y_ml,
    test_size=0.20,
    random_state=42
)

print("\nTRAINING DATA:", X_train.shape)
print("TEST DATA:", X_test.shape)


# Linear Regression

linear_model = LinearRegression()

linear_model.fit(
    X_train,
    y_train
)

linear_prediction = linear_model.predict(
    X_test
)

linear_mae = mean_absolute_error(
    y_test,
    linear_prediction
)

linear_mse = mean_squared_error(
    y_test,
    linear_prediction
)

linear_rmse = np.sqrt(
    linear_mse
)

linear_r2 = r2_score(
    y_test,
    linear_prediction
)

print("\nLINEAR REGRESSION")
print("MAE:", linear_mae)
print("MSE:", linear_mse)
print("RMSE:", linear_rmse)
print("R2 Score:", linear_r2)


# Random Forest

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

rf_model.fit(
    X_train,
    y_train
)

rf_prediction = rf_model.predict(
    X_test
)

rf_mae = mean_absolute_error(
    y_test,
    rf_prediction
)

rf_mse = mean_squared_error(
    y_test,
    rf_prediction
)

rf_rmse = np.sqrt(
    rf_mse
)

rf_r2 = r2_score(
    y_test,
    rf_prediction
)

print("\nRANDOM FOREST")
print("MAE:", rf_mae)
print("MSE:", rf_mse)
print("RMSE:", rf_rmse)
print("R2 Score:", rf_r2)


# Model Comparison

comparison = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Random Forest"
    ],
    "MAE": [
        linear_mae,
        rf_mae
    ],
    "MSE": [
        linear_mse,
        rf_mse
    ],
    "RMSE": [
        linear_rmse,
        rf_rmse
    ],
    "R2 Score": [
        linear_r2,
        rf_r2
    ]
})

print("\nMODEL COMPARISON")
print(comparison)


# Random Forest Feature Importance

importance = pd.DataFrame({
    "Feature": features,
    "Importance": rf_model.feature_importances_
})

importance = importance.sort_values(
    by="Importance",
    ascending=False
)

print("\nFEATURE IMPORTANCE")
print(importance)


plt.figure(figsize=(9, 5))

plt.bar(
    importance["Feature"],
    importance["Importance"]
)

plt.xlabel("Feature")
plt.ylabel("Importance")
plt.title("Random Forest Feature Importance")

plt.xticks(rotation=45)

plt.show()


# Actual vs Predicted

plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    linear_prediction,
    alpha=0.6
)

plt.xlabel("Actual G3")
plt.ylabel("Predicted G3")
plt.title("Actual vs Predicted Final Grade")

plt.grid(True)

plt.show()


# New Student Prediction

new_student = pd.DataFrame({
    "age": [17],
    "studytime": [2],
    "failures": [0],
    "absences": [3],
    "G1": [14],
    "G2": [15]
})

predicted_grade = linear_model.predict(
    new_student
)

print("\nNEW STUDENT PREDICTION")
print("Predicted G3:", predicted_grade[0])


# High Performers

high_performers = df[
    df["G3"] >= 15
]

print("\nHIGH PERFORMERS")
print("Number of High Performers:", len(high_performers))

print("\nHIGH PERFORMER AVERAGES")
print(
    high_performers[
        [
            "age",
            "studytime",
            "failures",
            "absences",
            "G1",
            "G2",
            "G3"
        ]
    ].mean()
)


# Ethical Analysis

print("\nETHICAL CONSIDERATIONS")

print("1. Student data should be kept private.")
print("2. Personal identifiers should not be exposed.")
print("3. Predictions should support teachers, not replace them.")
print("4. The model should not unfairly label students.")
print("5. Demographic and socioeconomic bias should be monitored.")
print("6. Correlation should not be treated as causation.")
print("7. Model limitations should be clearly communicated.")


print("\nMULTIVARIATE ANALYSIS AND MACHINE LEARNING COMPLETED SUCCESSFULLY")
```
